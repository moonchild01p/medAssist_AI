```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import pickle
import os

os.makedirs('models', exist_ok=True)
```

```python
# Maternal Health Risk Prediction
df1 = pd.read_csv('Maternal Health Risk Data Set.csv')

le = LabelEncoder()
df1['RiskLevel_enc'] = le.fit_transform(df1['RiskLevel'])  # low=1, mid=2, high=0

X1 = df1[['Age', 'SystolicBP', 'DiastolicBP', 'BS', 'BodyTemp', 'HeartRate']]
y1 = df1['RiskLevel_enc']

scaler1 = StandardScaler()
X1_scaled = scaler1.fit_transform(X1)

X1_train, X1_test, y1_train, y1_test = train_test_split(X1_scaled, y1, test_size=0.2, random_state=42)

model1 = RandomForestClassifier(n_estimators=100, random_state=42)
model1.fit(X1_train, y1_train)

print("=== Risque grossesse ===")
print(f"Accuracy: {accuracy_score(y1_test, model1.predict(X1_test)):.2f}")
print(classification_report(y1_test, model1.predict(X1_test), target_names=le.classes_))

pickle.dump(model1, open('models/model_grossesse.pkl', 'wb'))
pickle.dump(scaler1, open('models/scaler_grossesse.pkl', 'wb'))
pickle.dump(le, open('models/encoder_grossesse.pkl', 'wb'))

```

```python
# heart failure prediction
df2 = pd.read_csv('heart_failure_clinical_records_dataset.csv')

X2 = df2.drop('DEATH_EVENT', axis=1)
y2 = df2['DEATH_EVENT']

scaler2 = StandardScaler()
X2_scaled = scaler2.fit_transform(X2)

X2_train, X2_test, y2_train, y2_test = train_test_split(X2_scaled, y2, test_size=0.2, random_state=42)

model2 = GradientBoostingClassifier(n_estimators=100, random_state=42)
model2.fit(X2_train, y2_train)

print("=== Arrêt cardiaque ===")
print(f"Accuracy: {accuracy_score(y2_test, model2.predict(X2_test)):.2f}")
print(classification_report(y2_test, model2.predict(X2_test)))

pickle.dump(model2, open('models/model_cardiaque.pkl', 'wb'))
pickle.dump(scaler2, open('models/scaler_cardiaque.pkl', 'wb'))
```

```python
# score santé général

df3 = df1.copy()
df3['high_risk'] = (df3['RiskLevel'] == 'high risk').astype(int)

X3 = df3[['Age', 'SystolicBP', 'DiastolicBP', 'BS', 'BodyTemp', 'HeartRate']]
y3 = df3['high_risk']

scaler3 = StandardScaler()
X3_scaled = scaler3.fit_transform(X3)

X3_train, X3_test, y3_train, y3_test = train_test_split(X3_scaled, y3, test_size=0.2, random_state=42)

model3 = LogisticRegression(random_state=42)
model3.fit(X3_train, y3_train)

print("=== Score santé général ===")
print(f"Accuracy: {accuracy_score(y3_test, model3.predict(X3_test)):.2f}")
print(classification_report(y3_test, model3.predict(X3_test)))

pickle.dump(model3, open('models/model_sante.pkl', 'wb'))
pickle.dump(scaler3, open('models/scaler_sante.pkl', 'wb'))
```
